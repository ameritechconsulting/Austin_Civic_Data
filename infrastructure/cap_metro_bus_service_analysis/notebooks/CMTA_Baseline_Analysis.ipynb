# CMTA + City of Austin Transit Baseline Evidence

This notebook reproduces the baseline evidence workflow for transit infrastructure adequacy, safety, and accountability using public datasets already in the workspace.

## 1) Setup

```python
from pathlib import Path
import pandas as pd

root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
tables = root / 'output' / 'tables'
figures = root / 'output' / 'figures'

print('Root:', root)
print('Tables path exists:', tables.exists())
print('Figures path exists:', figures.exists())
```

## 2) Run Core Pipeline Scripts

```python
import subprocess

cmd1 = ['python3', str(root / 'scripts' / 'run_baseline_analysis.py')]
cmd2 = ['python3', str(root / 'scripts' / 'make_figures.py')]

print(subprocess.run(cmd1, capture_output=True, text=True).stdout)
print(subprocess.run(cmd2, capture_output=True, text=True).stdout)
```

## 3) Load Outputs

```python
evidence = pd.read_csv(tables / 'evidence_matrix.csv')
risk = pd.read_csv(tables / 'stop_baseline_risk_index.csv')
trends = pd.read_csv(tables / 'yearly_complaint_trends.csv')
responsibility = pd.read_csv(tables / 'jurisdiction_responsibility_summary.csv')

display(evidence)
display(risk.head(10))
display(trends)
display(responsibility.head(10))
```

## 4) Key Indicators

```python
metrics = evidence[['metric', 'value']].copy()
display(metrics)

print('Top 5 high-risk stops:')
display(risk[['stop_name', 'trip_count', 'total_calls', 'risk_score']].head(5))
```

## 5) View Generated Figures

```python
import matplotlib.pyplot as plt

for name in [
    'service_vs_complaints_bar.png',
    'complaint_trends_by_year_bar.png',
    'jurisdiction_split_top_stops.png'
]:
    p = figures / name
    if p.exists():
        print(name)
        img = plt.imread(p)
        plt.figure(figsize=(10, 6))
        plt.imshow(img)
        plt.axis('off')
        plt.title(name)
        plt.show()
    else:
        print('Missing:', name)
```
